```python
!pip install rdkit
import pandas as pd
from rdkit import Chem
from rdkit.Chem import Descriptors, Lipinski, Crippen, rdMolDescriptors
from rdkit.Chem.rdFingerprintGenerator import GetMorganGenerator
```

```python
# 1. Load your normalized CSV file
df = pd.read_csv("/content/sample_data/drug_featurization.csv")
```

```python
# 6. Generate 2048-bit Morgan fingerprints
fingerprint_generator = GetMorganGenerator(radius=2, fpSize=2048)

def make_fingerprint(smiles):
    mol = Chem.MolFromSmiles(smiles)
    fingerprint = fingerprint_generator.GetFingerprintAsNumPy(mol)
    return fingerprint

fingerprint_data = df["canonical_smiles"].apply(make_fingerprint)
```

```python
# Convert fingerprint values into columns: FP_0, FP_1, ..., FP_2047
fingerprint_df = pd.DataFrame(
    fingerprint_data.tolist(),
    columns=[f"FP_{i}" for i in range(2048)]
)
```

```python
# 7. Join descriptors and fingerprints
final_df = pd.concat(
    [df.reset_index(drop=True), fingerprint_df.reset_index(drop=True)],
    axis=1
)
```
